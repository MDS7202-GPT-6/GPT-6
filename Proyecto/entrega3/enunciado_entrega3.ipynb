#### Código de funciones

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

def evaluar_predicciones_universo_transacciones(pred_csv, real_parquet, universo_parquet="dataReal/transacciones.parquet"):
    """
    pred_csv: archivo CSV SIN header con columnas (client_id, product_id)
    real_parquet: archivo .parquet con columnas customer_id, product_id
    universo_parquet: archivo PARQUET con todos los pares posibles (TN y FN)
    """

    df_univ = pd.read_parquet(universo_parquet)[["customer_id", "product_id"]].drop_duplicates()
    df_real = pd.read_parquet(real_parquet)[["customer_id", "product_id"]].drop_duplicates()
    df_real["real"] = 1
    df_pred = pd.read_csv(pred_csv, header=None)
    df_pred.columns = ["customer_id", "product_id"]
    df_pred = df_pred.drop_duplicates()
    df_pred["pred"] = 1
    print("#"*10+"Tamaños de los datasets"+"#"*10)
    print("Tamaño universo (transacciones): ", len(df_univ))
    print("Tamaño real (transacciones): ", len(df_real))
    print("Tamaño predicción (transacciones): ", len(df_pred))
    print("#"*40)

    df_eval = (
        df_univ
        .merge(df_real, on=["customer_id", "product_id"], how="left")
        .merge(df_pred, on=["customer_id", "product_id"], how="left")
    )

    df_eval["real"] = df_eval["real"].fillna(0).astype(int)
    df_eval["pred"] = df_eval["pred"].fillna(0).astype(int)

    y_true = df_eval["real"]
    y_pred = df_eval["pred"]

    tn = ((y_true == 0) & (y_pred == 0)).sum()
    fp = ((y_true == 0) & (y_pred == 1)).sum()
    fn = ((y_true == 1) & (y_pred == 0)).sum()
    tp = ((y_true == 1) & (y_pred == 1)).sum()

    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    recall = recall_score(y_true, y_pred, zero_division=0)
    balanced_acc = (recall + specificity) / 2

    metrics = {
        "TP": tp,
        "FP": fp,
        "FN": fn,
        "TN": tn,
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred, zero_division=0),
        "recall": recall,
        "f1": f1_score(y_true, y_pred, zero_division=0),
        "specificity": specificity,
        "balanced_accuracy": balanced_acc,
    }

    print("==============================================")
    print("RESULTADOS DE LAS MÉTRICAS")
    print("==============================================")
    print(f"TP (Verdaderos Positivos):   {tp}")
    print(f"FP (Falsos Positivos):       {fp}")
    print(f"FN (Falsos Negativos):       {fn}")
    print(f"TN (Verdaderos Negativos):   {tn}")
    print("----------------------------------------------")
    print(f"Accuracy:          {metrics['accuracy']:.4f}")
    print(f"Precision:         {metrics['precision']:.4f}")
    print(f"Recall (TPR):      {metrics['recall']:.4f}")
    print(f"Specificity (TNR): {metrics['specificity']:.4f}")
    print(f"Balanced Accuracy: {metrics['balanced_accuracy']:.4f}")
    print(f"F1 Score:          {metrics['f1']:.4f}")
    print("==============================================\n")
    cm = pd.crosstab(df_eval["real"], df_eval["pred"], normalize="index")

    print("Matriz de Confusión (normalizada por fila):")
    print(cm, "\n")

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm, annot=True, fmt=".2f",
        cmap="Blues",
        vmin=0, vmax=1 
    )
    plt.title("Matriz de Confusión (Normalizada)")
    plt.xlabel("Predicción")
    plt.ylabel("Real")
    plt.show()

    return metrics
````

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

clientes_df = pd.read_parquet('dataReal/clientes.parquet')
productos_df = pd.read_parquet('dataReal/productos.parquet')

universe_size = len(clientes_df) * len(productos_df)

def evaluar_predicciones_universo(batch_id, predcsv):
    # Cargar datos reales
    data = pd.read_parquet(f"dataReal/batch_t{batch_id}.parquet")
    data = data.groupby(['customer_id', 'product_id']).size().reset_index()
    true_pairs = set(map(tuple, data[['customer_id', 'product_id']].values))

    # Cargar predicciones
    #preds = pd.read_csv(f"predictions/future_pairsBatch{batch_id-1}.csv",
    #                    names=['customer_id', 'product_id'])
    preds = pd.read_csv(predcsv,
                        names=['customer_id', 'product_id'])
    pred_pairs = set(map(tuple, preds[['customer_id', 'product_id']].values))
    print("Predicciones realizadas: ", len(pred_pairs))
    print("universo: ", universe_size)
    # TP = predicho y real
    tp = len(pred_pairs & true_pairs)
    # FP = predicho pero no real
    fp = len(pred_pairs - true_pairs)
    # FN = real pero no predicho
    fn = len(true_pairs - pred_pairs)
    # TN = universo total - todo lo demás
    tn = universe_size - tp - fp - fn

    y_true = [1]*tp + [0]*fp + [1]*fn + [0]*tn
    y_pred = [1]*tp + [1]*fp + [0]*fn + [0]*tn
    specificity = tn / (tn + fp) if (tn + fp) > 0 else 0
    recall = recall_score(y_true, y_pred, zero_division=0)
    balanced_acc = (recall + specificity) / 2

    print("-"*50)
    print("compras efectivas: "+ str(tp+fn))
    print("porcentaje de compras efectivas recomendadas: "+ str(tp/(tp+fn)*100))
    metrics = {
        "accuracy": accuracy_score(y_true, y_pred),
        "precision": precision_score(y_true, y_pred),
        "recall": recall_score(y_true, y_pred),
        "f1": f1_score(y_true, y_pred),
        "specificity": specificity,
        "balanced_accuracy": balanced_acc,
    }

    print("==============================================")
    print("RESULTADOS DE LAS MÉTRICAS")
    print("==============================================")
    print(f"TP (Verdaderos Positivos): {tp}")
    print(f"FP (Falsos Positivos):     {fp}")
    print(f"FN (Falsos Negativos):     {fn}")
    print(f"TN (Verdaderos Negativos): {tn}")
    print("----------------------------------------------")
    print(f"Accuracy:          {metrics['accuracy']:.4f}")
    print(f"Precision:         {metrics['precision']:.4f}")
    print(f"Recall (TPR):      {metrics['recall']:.4f}")
    print(f"Specificity (TNR): {metrics['specificity']:.4f}")
    print(f"Balanced Accuracy: {metrics['balanced_accuracy']:.4f}")
    print(f"F1 Score:          {metrics['f1']:.4f}")
    print("==============================================\n")

    cm =  pd.crosstab(pd.Series(y_true, name="Real"),
                 pd.Series(y_pred, name="Pred"),
                 normalize="index")

    print("Matriz de Confusión (normalizada por fila):")
    print(cm, "\n")

    plt.figure(figsize=(6, 5))
    sns.heatmap(
        cm, annot=True, fmt=".2f",
        cmap="Blues",
        vmin=0, vmax=1  
    )
    plt.title("Matriz de Confusión (Normalizada)")
    plt.xlabel("Predicción")
    plt.ylabel("Real")
    plt.show()

    return metrics
````

````python
def comparar_clientes(batch_id, predcsv):
    import pandas as pd
    import matplotlib.pyplot as plt
    import seaborn as sns
    
    # Leer predicciones
    preds = pd.read_csv(predcsv, names=['customer_id', 'product_id'])
    pred_counts = preds['customer_id'].value_counts().reset_index()
    pred_counts.columns = ['customer_id', 'pred_count']
    
    # Leer batch real
    batch_data = pd.read_parquet(f"dataReal/batch_t{batch_id}.parquet")
    true_counts = batch_data['customer_id'].value_counts().reset_index()
    true_counts.columns = ['customer_id', 'true_count']
    
    # Unir
    merged_counts = pred_counts.merge(true_counts, on='customer_id', how='outer').fillna(0)

    # -------- PLOT: DOS HISTOGRAMAS -------- #
    plt.figure(figsize=(12, 6))

    sns.histplot(merged_counts['true_count'], 
                 color='blue', 
                 label='Compras reales', 
                 alpha=0.5, 
                 bins=30)

    sns.histplot(merged_counts['pred_count'], 
                 color='orange', 
                 label='Predicciones', 
                 alpha=0.5, 
                 bins=30)

    plt.title(f'Histograma Comparativo: Compras Reales vs Predicciones (Batch {batch_id})')
    plt.xlabel('Número de productos por cliente')
    plt.ylabel('Cantidad de clientes')
    plt.legend()
    plt.tight_layout()
    plt.show()
````

Uno de los primeros hallazgos que se lograron obtener, es que el universo en la evaluación/generación de los modelos en los entregables anteriores no consideraba la totalidad: Se había calculado el universo según transacciones previas, pero no se había incluido la totalidad de clientes y productos posibles. 

Al hacer el dataset utilizado para el entrenamiento, evaluación y test del modelo se consideraron los datos de las transacciones para generar los pares cliente-producto posibles, en lugar de considerar todos los clientes y productos del catálogo. Esto porque el dataset resultante era mucho más manejable en tamaño y permitía entrenar modelos de forma más eficiente. Sin embargo, al momento de evaluar el desempeño del modelo en un entorno real, este enfoque resultó en una subestimación del universo real de posibles interacciones cliente-producto.

Por ello, al optimizar el desempeño del modelo para la competencia, se realizó según el f1 score pero considerando el universo generado según las transacciones históricas. Ahora en el informe se usará el universo completo (todos los clientes y productos posibles) para evaluar el desempeño del modelo en un entorno más realista y con las mismas métricas utilizadas en el concurso.

Para poder evaluar el desempeño de los modelos, se crearon funciones que calculan diferentes métricas en cada semana de forma individual. Lo que varía entre estas funciones es la forma de calcular el universo.

Estas funciones toman como entrada las predicciones generadas por el modelo y los valores reales, y calcularán las siguientes métricas:
- TP (Verdaderos Positivos)
- FP (Falsos Positivos)
- FN (Falsos Negativos)
- TN (Verdaderos Negativos)
- Accuracy
- Precision
- Recall (TPR)
- Specificity (TNR)
- Balanced Accuracy
- F1 Score (binario)

Estas métricas nos permiten evaluar las proporciones de aciertos y errores de distintos tipos a lo largo de cada semana de forma individual, por lo que permiten un análisis detallado del desempeño del modelo. Se mostrará también una matriz de confusión normalizada.

### 📌 Análisis comparativo [2.0 puntos]

#### Código de generación de métricas y gráficos comparativos

````python
# código para realizar análisis de resultados entre batchs
def compare_results(res1,res2,res3,res4):
    resultados = [res1, res2, res3, res4]
    df_res = pd.DataFrame(resultados)
    df_res["batch"] = ["Batch 1", "Batch 2", "Batch 3", "Batch 4"]
    print("#"*10+" Métricas comparativas entre batches "+"#"*10)
    print(df_res)
    print("#"*57)
    plt.figure(figsize=(8,4))
    sns.barplot(data=df_res, x="batch", y="f1", palette="Blues")
    plt.title("F1 Score por Batch")
    plt.ylim(0,1)
    plt.show()

    df_melt = df_res.melt(id_vars="batch", value_vars=["precision", "recall", "specificity"])
    plt.figure(figsize=(10,4))
    sns.barplot(data=df_melt, x="batch", y="value", hue="variable", palette="Set2")
    plt.title("Métricas por Batch")
    plt.ylim(0,1)
    plt.show()
````

#### Resultados comparativos entre batches y análisis

````python
compare_results(resultado1_valido, resultado2_valido, resultado3_valido, resultado4_valido)
````
