En el competitivo universo de las bebidas gaseosas, la empresa **SodAI Drinks 🥤** ha logrado destacarse por su creatividad, diversidad de productos y enfoque centrado en el cliente. Ofrece una extensa gama de bebidas carbonatadas que abarca distintos segmentos del mercado: desde productos premium en presentaciones sofisticadas, hasta gaseosas accesibles para el consumo masivo, disponibles en diversos tamaños y tipos de envases. 

La compañía opera en múltiples regiones y zonas, sirviendo a una variedad de puntos de venta que incluyen desde tiendas de conveniencia y minimarkets hasta el canal frío tradicional. Cada tipo de cliente tiene sus particularidades: algunos reciben entregas hasta 4 veces por semana, mientras que otros son visitados por la fuerza de ventas solo una vez semanalmente. Esta diversidad de perfiles representa tanto una oportunidad como un desafío comercial: ¿cómo saber qué productos tienen más chances de ser comprados por cada cliente en un momento dado?

Con el objetivo de aumentar la facturación de forma inteligente y mejorar la eficiencia de su estrategia de ventas, **SodAI Drinks** decide crear una nueva célula interna de innovación: el equipo **Deep Drinkers 🤖**, cuyo propósito es aplicar ciencia de datos para anticiparse a las necesidades del cliente y potenciar el negocio desde una perspectiva basada en información.

El corazón de esta iniciativa es el desarrollo de un sistema predictivo personalizado para cada cliente. Para ello, **Deep Drinkers** convoca a un equipo de Data Scientists y especialistas en *machine learning* con una misión clara: construir un modelo predictivo que, cada semana, pueda estimar la probabilidad de compra de cada producto del portafolio para cada cliente activo.

El modelo deberá tener en cuenta múltiples factores, incluyendo:
- **Tipo de cliente**, ej. "TIENDA DE CONVENIENCIA", "MINIMARKET".
- **Frecuencia de entregas y visitas**, indicadores del nivel de actividad comercial.
- **Ubicación geográfica** (por región y zona).
- **Preferencias históricas de consumo**, inferidas por patrones de compra anteriores.
- **Características del producto**, como marca, categoría, segmento, tipo de envase y tamaño

El objetivo final es que, **cada semana**, se genere una tabla de productos priorizados: para cada cliente, un listado de productos ordenado por su probabilidad estimada de compra. Esta información será enviada al equipo comercial, que podrá usarla en call center, para incrementar las chances de concretar ventas al ofrecer justo lo que el cliente probablemente quiere comprar.

Este proyecto representa un cambio de paradigma en la forma en que **SodAI Drinks** gestiona su fuerza de ventas: de un enfoque reactivo y generalista, a uno proactivo, basado en datos y profundamente personalizado. Así, la empresa no solo espera aumentar su rentabilidad, sino también construir relaciones más sólidas con sus clientes, ofreciéndoles recomendaciones más relevantes y oportunas.

Para lograr lo anterior, el equipo **Deep Drinkers** contará con los siguientes conjuntos de datos, junto a sus respectivos atributos:

- **Datos transaccionales** (`transacciones.parquet`): contiene el historial de compras realizadas por los clientes.

	- `package`: tipo de envase del producto.
	- `size`: tamaño del producto en litros.

## Cruce de información

En el EDA al estudiar "Items", nos dimos cuenta que habían valores en negativos. Estos al estudiar el tipo de compra con los demás datos, nos dimos cuenta que habían filas iguales (customer_id, product_id y semana) lo que sugiere que pueden ser transacciones que se devolvieron o cancelaron. Por esto, se decide "sumar" los items con las mismas características para que se mantenga la información de los productos que realmente fueron comprados.

```python
# agrupamos por las variables identficadoras de transaccion
df_transacciones = df_transacciones.groupby(
    ["order_id", "product_id", "customer_id", "purchase_date"],
    as_index=False
)["items"].sum() #luego, sumamos los items
# Mantenemos los items mayores a 0, ya que en el EDA luego de hacer lo anterior aún habían
# algunos items con valor negativo, sospechamos que podrían ser errores y como son pocos casos, se deciden filtrar.
df_transacciones = df_transacciones[df_transacciones["items"] >= 0] 

# Creamos las variable semana y año
df_transacciones["Semana"] = df_transacciones["purchase_date"].dt.isocalendar().week
df_transacciones["Año"] = df_transacciones["purchase_date"].dt.year

# Creamos el target, como solo tenemos las transacciones realizadas, de momento es = 1.
df_transacciones["target"] = 1


# obtenemos los valores únicos
clientes = df_transacciones["customer_id"].unique()
productos = df_transacciones["product_id"].unique()
semanas = df_transacciones[["Año", "Semana"]].drop_duplicates()

todasCombinaciones = ( #generamos las posibles combinaciones
    pd.DataFrame({"customer_id": clientes})
    .merge(pd.DataFrame({"product_id": productos}), how="cross")
    .merge(semanas, how="cross")
)
# Hacemos merge para generar filas que nos den todas las combinaciones y establecer luego el target = 0.
df = todasCombinaciones.merge(
    df_transacciones[["customer_id", "product_id","Año", "Semana", "target"]],
    on=["customer_id", "product_id", "Año", "Semana"],
    how="left"
)

df["target"] = df["target"].fillna(0).astype(int)
df = df.merge(df_clientes,on="customer_id").merge(df_productos,on="product_id")
```

```python

def validar_clientes(df_clientes):
    print("=== Validación Clientes ===")

    coords_invalidas = df_clientes[
        (df_clientes['Y'] < -90) | (df_clientes['Y'] > 90) |
        (df_clientes['X'] < -180) | (df_clientes['X'] > 180)
    ]
    print("Clientes con coordenadas inválidas:", len(coords_invalidas))
    
    # Frecuencias negativas
    entregas_invalidas = df_clientes[(df_clientes['num_deliver_per_week'] < 0) ]
    print("Clientes con entregas negativas:", len(entregas_invalidas))

    # Frecuencias negativas
    entregas_invalidas = df_clientes[(df_clientes['num_visit_per_week'] < 0)]
    print("Clientes con visitas negativas:", len(entregas_invalidas))

    # Verificar si algún zone_id está asociado a más de un region_id
    zonas_por_region = df_clientes.groupby("zone_id")["region_id"].nunique().reset_index(name="num_regiones")

    zonas_conflictivas = zonas_por_region[zonas_por_region["num_regiones"] > 1]

    print("Zonas asociadas a múltiples regiones:")
    print(zonas_conflictivas)

    if not zonas_conflictivas.empty:
        detalle_conflictivas = df_clientes[df_clientes["zone_id"].isin(zonas_conflictivas["zone_id"])]
        print("\nDetalle de zonas conflictivas:")
        print(detalle_conflictivas.sort_values(["zone_id", "region_id"]))
    
       

def validar_productos(df_productos):
    print("\n=== Validación Productos ===")
    tamanos_invalidos = df_productos[df_productos['size'] <= 0]
    print("Productos con tamaño inválido:", len(tamanos_invalidos))
    # Verificar si alguna sub_category está asociada a más de una category
    subcat_por_cat = (
        df_productos.groupby("sub_category")["category"]
        .nunique()
        .reset_index(name="num_categorias")
    )

    # Filtrar subcategorías conflictivas
    subcat_conflictivas = subcat_por_cat[subcat_por_cat["num_categorias"] > 1]

    print("Subcategorías asociadas a múltiples categorías:")
    print(subcat_conflictivas)

    if not subcat_conflictivas.empty:
        detalle_conflictivas = df_productos[
            df_productos["sub_category"].isin(subcat_conflictivas["sub_category"])
        ]
        print("\nDetalle de subcategorías conflictivas:")
        print(detalle_conflictivas.sort_values(["sub_category", "category"]))

def validar_transacciones(df_transacciones):
    print("\n=== Validación Transacciones ===") 
    # Cantidades negativas o cero
    items_invalidos = df_transacciones[df_transacciones['items'] <= 0]
    print("Transacciones con items inválidos:", len(items_invalidos))
    



validar_clientes(df_clientes)
validar_productos(df_productos)
validar_transacciones(df_transacciones)
```

En esta sección, se deben construir pipelines para automatizar el preprocesamiento de los datos, lo cual garantizará que el flujo de trabajo sea reproducible y eficiente para esta entrega y las futuras. El objetivo es aplicar una serie de transformaciones en un orden definido para asegurar que los datos estén listos para los modelos a entrenar. El pipeline final debe incluir las técnicas de pre-procesamiento que se deben aplicar a los distintos datos (según lo que consideren necesario para el problema). Por ejemplo:

- **Imputación de valores nulos**: Manejo de datos faltantes mediante estrategias adecuadas (media, mediana, moda, interpolación, etc.). 

- **Transformaciones personalizadas**: Uso de ColumnTransformer para aplicar diferentes transformaciones a columnas específicas.

- **Codificación de variables categóricas**: Convertir datos categóricos a un formato numérico adecuado (One-Hot Encoding, Label Encoding, etc.).

- **Discretización de variables**: Conversión de variables numéricas continuas en categorías si son relevantes para el desempeño del modelo a entrenar.

- **Estandarización o normalización** : Ajustar la escala de los datos para mejorar el rendimiento de los algoritmos sensibles a la magnitud de las variables.

- **Eliminación o transformación de valores atípicos**: Identificar y tratar con datos outliers para mejorar la robustez del modelo.

- **Nuevas características** : Creación de variables adicionales que puedan aportar información relevante al modelo.

Cada una de estas transformaciones debe ser justificada en función de su relevancia para el problema y los datos, y es importante evaluar su impacto en el rendimiento del modelo. Además, el pipeline debe ser flexible y modular para poder probar diferentes configuraciones de preprocesamiento.

```python
import pandas as pd
import numpy as np
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.cluster import KMeans
# Creamos una variable que indique un cluster al que pertenece el cliente según su ubicación geográfica
class GeoClustering(BaseEstimator, TransformerMixin):
    def __init__(self, n_clusters=4, random_state=42):
        self.n_clusters = n_clusters
        self.random_state = random_state
        self.kmeans = None

    def fit(self, X, y=None):
        if "X" in X.columns and "Y" in X.columns:
            coords = X[["X", "Y"]].dropna()
            self.kmeans = KMeans(n_clusters=self.n_clusters, random_state=self.random_state, n_init=10)
            self.kmeans.fit(coords)
        return self

    def transform(self, X):
        X = X.copy()
        if self.kmeans is not None and "X" in X.columns and "Y" in X.columns:
            mask = X["X"].notna() & X["Y"].notna()
            X.loc[mask, "geo_cluster"] = self.kmeans.predict(X.loc[mask, ["X", "Y"]])
            X["geo_cluster"] = X["geo_cluster"].fillna(-1).astype(int)
        else:
            X["geo_cluster"] = -1
        #print("Listo el cluster compañero")
        #print(X.head())
        return X

# Clase creada para eliminar los outliers usando estrategia IQR
class IQR(BaseEstimator, TransformerMixin):
    def __init__(self, l=1.5):
        self.l = l
    def fit(self, X, y=None):
        q1 = X.quantile(0.25)
        q3 = X.quantile(0.75)
        iqr = q3 - q1
        self.inferior = q1 - self.l * iqr
        self.superior = q3 + self.l * iqr
        return self
    def transform(self, X):
        X = X.copy()
        for col in X.columns:
            X.loc[(X[col] < self.inferior[col]) | (X[col] > self.superior[col]), col] = np.nan
        return X

# clase creada para limpiar los datos
class CustomCleaner(BaseEstimator, TransformerMixin):
    def fit(self, X, y=None): return self
    def transform(self, X):
        X = X.copy()
        for col in X.select_dtypes(include=["int64", "float64"]).columns:
            X[col] = X[col].mask(X[col] <= 0, np.nan)
        for col in ["X", "Y"]:
            if col in X.columns:
                if col == "X":
                    X[col] = X[col].mask((X[col] < -180) | (X[col] > 180))
                elif col == "Y":
                    X[col] = X[col].mask((X[col] < -90) | (X[col] > 90))
        X = X.drop(columns=[col for col in ["X", "Y"] if col in X.columns])
        return X

# Aquí generamos los datos: 
# - Frecuencia: Cuántas veces cierto cliente compró el producto previamente a partir de la fecha actual que se está mirando (fila)
# - Trimestre: Indique el trimestre en que se hizo la compra -> así tenemos data temporal manejable con categorías
# - Frecuencia por brand: similar a frecuencia, pero en vez de por producto, por marca.
class FeatureAggregator(BaseEstimator, TransformerMixin):
    def __init__(self):
        pass

    def fit(self, X, y=None):
        return self

    def transform(self, X, df_transacciones=None):
        if df_transacciones is None:
            raise ValueError("Necesitas pasar df_transacciones a transform()")
        
        df_transacciones = df_transacciones.copy()
        df_transacciones["purchase_date"] = pd.to_datetime(df_transacciones["purchase_date"])

        X = X.copy()
        X["fecha_actual"] = pd.to_datetime(
            X["Año"].astype(str) + "-W" + X["Semana"].astype(str) + "-1",
            format="%G-W%V-%u"
        )
        trans = df_transacciones.rename(columns={"purchase_date": "fecha_compra"})
        merged = X.merge(trans, on=["customer_id", "product_id"], how="left")
        merged = merged.rename(columns={"Año_x": "Año","Semana_x":"Semana"})

        merged = merged[merged["fecha_compra"] <= merged["fecha_actual"]]

        freq = (
            merged.groupby(["customer_id", "product_id", "Año", "Semana"])
            .size()
            .reset_index(name="frecuencia")
        )
        X = X.merge(freq, on=["customer_id", "product_id", "Año", "Semana"], how="left")
        X["frecuencia"] = X["frecuencia"].fillna(0).astype(int)

        X["trimestre"] = X["fecha_actual"].dt.quarter

        if "brand" in merged.columns:
            freq_brand = (
                merged.groupby(["customer_id", "brand", "Año", "Semana"])
                .size()
                .reset_index(name="frecuencia_brand")
            )
            X = X.merge(freq_brand, on=["customer_id", "brand", "Año", "Semana"], how="left")
            X["frecuencia_brand"] = X["frecuencia_brand"].fillna(0).astype(int)

        X = X.drop(columns=["fecha_actual"])
        return X
    
"""
class FeatureAggregator(BaseEstimator, TransformerMixin):
    def __init__(self, df_transacciones):
        self.df_transacciones = df_transacciones.copy()
        self.df_transacciones["purchase_date"] = pd.to_datetime(self.df_transacciones["purchase_date"])

    def fit(self, X, y=None):
        return self  # No necesita fit

    def transform(self, X):
        X = X.copy()
        X["fecha_actual"] = pd.to_datetime(
            X["Año"].astype(str) + "-W" + X["Semana"].astype(str) + "-1",
            format="%G-W%V-%u"
        )

        # Merge con transacciones
        trans = self.df_transacciones.rename(columns={"purchase_date": "fecha_compra"})
        merged = X.merge(trans, on=["customer_id", "product_id"], how="left")
        merged = merged.rename(columns={"Año_x": "Año","Semana_x":"Semana"})

    
        # Filtrar solo transacciones hasta la fecha actual
        merged = merged[merged["fecha_compra"] <= merged["fecha_actual"]]

        # Contar frecuencia por fila
        freq = (
            merged.groupby(["customer_id", "product_id", "Año", "Semana"])
            .size()
            .reset_index(name="frecuencia")
        )

        # Merge de vuelta con X
        X = X.merge(freq, on=["customer_id", "product_id", "Año", "Semana"], how="left")
        X["frecuencia"] = X["frecuencia"].fillna(0).astype(int)

        # Limpiar columnas temporales
        X = X.drop(columns=["fecha_actual"])
        #print(X.head())
        return X

"""




# definimos las columnas a trabajar
numeric_features = ["num_deliver_per_week", "num_visit_per_week", "size", "frecuencia", "promedio_cantidad", "recencia"]
categorical_features = ["customer_type", "brand", "category", "sub_category", "segment", "package"]

numeric_features = ["num_deliver_per_week", "num_visit_per_week", "size","frecuencia","frecuencia_brand"]
categorical_features = ["customer_type", "brand", "category", "segment", "package","geo_cluster", "trimestre"]
# definimos los pipelines
numeric_pipeline = Pipeline([
    ("iqr", IQR()),
    #("clean", CustomCleaner()),
    ("imputer", SimpleImputer(strategy="most_frequent",fill_value=0)),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

# Definimos el preprocesador
preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

from sklearn.preprocessing import FunctionTransformer

def agregar_features(X):
    return FeatureAggregator().transform(X, df_transacciones=df_transacciones)

pipeline_pp = Pipeline([
    ("features", FunctionTransformer(agregar_features)),
    ("geo_cluster", GeoClustering(n_clusters=4)),
    ("preprocessing", preprocessor)
])
# Separar X e y 
X_train = train_df.drop(columns=["target"]) 
y_train = train_df["target"] 
y_val = val_df["target"] 
y_test = test_df["target"] 
```

Como vimos previamente en el EDA, el dataset está extremadamente desbalanceado. Por ello, decidimos hacer un undersampling de la clase dominante (0), ya que los resultados obtenidos estaban siendo muy desfavorables para la clase 1.

```python
from sklearn.pipeline import Pipeline
from sklearn.dummy import DummyClassifier
# undersampling fuera del pipeline, en el pipeline da muchos problemas
# indices balanceados
def underSamp(Xtrain,ytrain):
    rng = np.random.RandomState(42)
    idx_class0 = np.where(ytrain == 0)[0]
    idx_class1 = np.where(ytrain == 1)[0]
    n_min = min(len(idx_class0), len(idx_class1))

    idx0_sub = rng.choice(idx_class0, n_min, replace=False)
    idx1_sub = rng.choice(idx_class1, n_min, replace=False)
    idx_final = np.concatenate([idx0_sub, idx1_sub])
    rng.shuffle(idx_final)
    return Xtrain.iloc[idx_final], ytrain.iloc[idx_final]

X_bal, y_bal = underSamp(X_train,y_train)

full_pipeline = Pipeline([
    ("preprocessing", pipeline_pp),
    ("classifier", DummyClassifier(strategy="most_frequent"))
])

full_pipeline.fit(X_bal, y_bal)
```

A partir de su análisis anterior, se debe proceder a optimizar el rendimiento del modelo seleccionado mediante la optimización de sus hiperparámetros. Para ello, se espera que implementen `Optuna` para optimizar no solo los hiperparámetros del modelo, sino también los de los preprocesadores utilizados (por ejemplo, OneHot Encoding, Scalers, etc.).

Al desarrollar este proceso, deberán responder las siguientes preguntas clave como mínimo:

```python
import optuna
from sklearn.model_selection import cross_val_score
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import make_scorer, f1_score

# Usamos F1-score macro porque el dataset está desbalanceado
scorer = make_scorer(f1_score, average="macro")

def objective(trial):
    #hiperparámetros
    max_depth = trial.suggest_int("max_depth", 2, 30)
    min_samples_split = trial.suggest_int("min_samples_split", 2, 20)
    min_samples_leaf = trial.suggest_int("min_samples_leaf", 1, 10)
    criterion = trial.suggest_categorical("criterion", ["gini", "entropy", "log_loss"])

    # Modelo dentro del pipeline de preprocesamiento
    clf = Pipeline([
        ("preprocessing", pipeline_pp),
        ("classifier", DecisionTreeClassifier(
            max_depth=max_depth,
            min_samples_split=min_samples_split,
            min_samples_leaf=min_samples_leaf,
            criterion=criterion,
            random_state=42
        ))
    ])

    # Validación cruzada en training set
    scores = cross_val_score(clf, X_bal, y_bal, scoring=scorer, cv=3, n_jobs=-1)
    return scores.mean()

# Crear y ejecutar el estudio
study = optuna.create_study(direction="maximize")
study.optimize(objective, n_trials=30, n_jobs=1)

print("Mejores hiperparámetros:", study.best_params)
print("Mejor F1 (cv):", study.best_value)
```

```python
best_params = study.best_params

best_model = Pipeline([
    ("preprocessing", pipeline_pp),
    ("classifier", DecisionTreeClassifier(
        **best_params,
        random_state=42
    ))
])

best_model.fit(X_bal, y_bal)
y_pred = best_model.predict(X_val)

print("Reporte en test con mejores hiperparámetros:")
print(classification_report(y_val, y_pred, digits=4))
```

Decidimos usar F1-score (macro) como métrica a optimizar, porque el dataset está desbalanceado y queremos que el modelo aprenda tanto de la clase mayoritaria como de la minoritaria.
