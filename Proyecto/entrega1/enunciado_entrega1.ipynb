En el competitivo universo de las bebidas gaseosas, la empresa **SodAI Drinks 🥤** ha logrado destacarse por su creatividad, diversidad de productos y enfoque centrado en el cliente. Ofrece una extensa gama de bebidas carbonatadas que abarca distintos segmentos del mercado: desde productos premium en presentaciones sofisticadas, hasta gaseosas accesibles para el consumo masivo, disponibles en diversos tamaños y tipos de envases. 

La compañía opera en múltiples regiones y zonas, sirviendo a una variedad de puntos de venta que incluyen desde tiendas de conveniencia y minimarkets hasta el canal frío tradicional. Cada tipo de cliente tiene sus particularidades: algunos reciben entregas hasta 4 veces por semana, mientras que otros son visitados por la fuerza de ventas solo una vez semanalmente. Esta diversidad de perfiles representa tanto una oportunidad como un desafío comercial: ¿cómo saber qué productos tienen más chances de ser comprados por cada cliente en un momento dado?

Con el objetivo de aumentar la facturación de forma inteligente y mejorar la eficiencia de su estrategia de ventas, **SodAI Drinks** decide crear una nueva célula interna de innovación: el equipo **Deep Drinkers 🤖**, cuyo propósito es aplicar ciencia de datos para anticiparse a las necesidades del cliente y potenciar el negocio desde una perspectiva basada en información.

El corazón de esta iniciativa es el desarrollo de un sistema predictivo personalizado para cada cliente. Para ello, **Deep Drinkers** convoca a un equipo de Data Scientists y especialistas en *machine learning* con una misión clara: construir un modelo predictivo que, cada semana, pueda estimar la probabilidad de compra de cada producto del portafolio para cada cliente activo.

El modelo deberá tener en cuenta múltiples factores, incluyendo:
- **Tipo de cliente**, ej. "TIENDA DE CONVENIENCIA", "MINIMARKET".
- **Frecuencia de entregas y visitas**, indicadores del nivel de actividad comercial.
- **Ubicación geográfica** (por región y zona).
- **Preferencias históricas de consumo**, inferidas por patrones de compra anteriores.
- **Características del producto**, como marca, categoría, segmento, tipo de envase y tamaño

El objetivo final es que, **cada semana**, se genere una tabla de productos priorizados: para cada cliente, un listado de productos ordenado por su probabilidad estimada de compra. Esta información será enviada al equipo comercial, que podrá usarla en call center, para incrementar las chances de concretar ventas al ofrecer justo lo que el cliente probablemente quiere comprar.

Este proyecto representa un cambio de paradigma en la forma en que **SodAI Drinks** gestiona su fuerza de ventas: de un enfoque reactivo y generalista, a uno proactivo, basado en datos y profundamente personalizado. Así, la empresa no solo espera aumentar su rentabilidad, sino también construir relaciones más sólidas con sus clientes, ofreciéndoles recomendaciones más relevantes y oportunas.

Para lograr lo anterior, el equipo **Deep Drinkers** contará con los siguientes conjuntos de datos, junto a sus respectivos atributos:

- **Datos transaccionales** (`transacciones.parquet`): contiene el historial de compras realizadas por los clientes.
	- `customer_id`: identificador único del cliente que realizó la compra.
	- `product_id`: identificador único del producto comprado.
	- `purchase_date`: fecha en que se realizó la transacción.
	- `order_id`: identificar de la orden de su pedido.
	- `payment`	monto total pagado por la transacción.

- **Datos de clientes** (`clientes.parquet`): incluye las características de cada cliente.
	- `customer_id`: identificador único del cliente.
	- `region_id`: identificador de la región geográfica donde se encuentra el cliente.
	- `customer_type`: tipo de cliente según el canal comercial, por ejemplo, “TIENDA DE CONVENIENCIA”.
	- `Y`: coordenada geográfica de latitud.
	- `X`: coordenada geográfica de longitud.
	- `num_deliver_per_week`: cantidad de entregas semanales que recibe el cliente.
	- `num_visit_per_week`: frecuencia de visitas de la fuerza de ventas por semana.

- **Datos de productos** (`productos.parquet`): describe las características de los productos del portafolio.
	- `product_id`: identificador único del producto.
	- `brand`: marca comercial del producto.
	- `category`: categoría general del producto, como “BEBIDAS CARBONATADAS”.
	- `sub_category`: subcategoría dentro de la categoría principal, por ejemplo, “GASEOSAS”.
	- `segment`: segmento de mercado al que pertenece el producto, como “PREMIUM”.
	- `package`: tipo de envase del producto.
	- `size`: tamaño del producto en litros.

```python

def validar_clientes(df_clientes):
    print("=== Validación Clientes ===")

    coords_invalidas = df_clientes[
        (df_clientes['Y'] < -90) | (df_clientes['Y'] > 90) |
        (df_clientes['X'] < -180) | (df_clientes['X'] > 180)
    ]
    print("Clientes con coordenadas inválidas:", len(coords_invalidas))
    
    # Frecuencias negativas
    entregas_invalidas = df_clientes[(df_clientes['num_deliver_per_week'] < 0) ]
    print("Clientes con entregas negativas:", len(entregas_invalidas))

    # Frecuencias negativas
    entregas_invalidas = df_clientes[(df_clientes['num_visit_per_week'] < 0)]
    print("Clientes con visitas negativas:", len(entregas_invalidas))

    # Verificar si algún zone_id está asociado a más de un region_id
    zonas_por_region = df_clientes.groupby("zone_id")["region_id"].nunique().reset_index(name="num_regiones")

    # Filtrar zonas conflictivas (más de una región asociada)
    zonas_conflictivas = zonas_por_region[zonas_por_region["num_regiones"] > 1]

    print("Zonas asociadas a múltiples regiones:")
    print(zonas_conflictivas)

    if not zonas_conflictivas.empty:
        detalle_conflictivas = df_clientes[df_clientes["zone_id"].isin(zonas_conflictivas["zone_id"])]
        print("\nDetalle de zonas conflictivas:")
        print(detalle_conflictivas.sort_values(["zone_id", "region_id"]))
    
       

def validar_productos(df_productos):
    print("\n=== Validación Productos ===")
    tamanos_invalidos = df_productos[df_productos['size'] <= 0]
    print("Productos con tamaño inválido:", len(tamanos_invalidos))
    # Verificar si alguna sub_category está asociada a más de una category
    subcat_por_cat = (
        df_productos.groupby("sub_category")["category"]
        .nunique()
        .reset_index(name="num_categorias")
    )

    # Filtrar subcategorías conflictivas
    subcat_conflictivas = subcat_por_cat[subcat_por_cat["num_categorias"] > 1]

    print("Subcategorías asociadas a múltiples categorías:")
    print(subcat_conflictivas)

    # Opcional: ver detalle de esas subcategorías conflictivas
    if not subcat_conflictivas.empty:
        detalle_conflictivas = df_productos[
            df_productos["sub_category"].isin(subcat_conflictivas["sub_category"])
        ]
        print("\nDetalle de subcategorías conflictivas:")
        print(detalle_conflictivas.sort_values(["sub_category", "category"]))

def validar_transacciones(df_transacciones):
    print("\n=== Validación Transacciones ===") 
    # Cantidades negativas o cero
    items_invalidos = df_transacciones[df_transacciones['items'] <= 0]
    print("Transacciones con items inválidos:", len(items_invalidos))
    



validar_clientes(df_clientes)
validar_productos(df_productos)
validar_transacciones(df_transacciones)
```

## Patrones de compra

```python
# Crear columna de año-semana
df['year_week'] = df['purchase_date'].dt.strftime('%Y-%U')

# Calcular productos comprados por cliente y semana
productos_por_semana = df.groupby(['customer_id', 'year_week'])['product_id'].nunique().reset_index(name='productos en semana')

# Promedio de productos por cliente (en todas sus semanas activas)
promedio_productos_cliente = productos_por_semana.groupby('customer_id')['productos en semana'].mean().reset_index(name='Promedio Prod por semana')

promedio_productos_cliente
```

Como son muchos clientes, haremos un histograma con plotly para visualizar la distribución de la cantidad de productos por semana.

```python
import plotly.express as px

fig = px.histogram(
    promedio_productos_cliente,
    x="Promedio Prod por semana",
    nbins=30,  # número de bins (ajusta según prefieras)
    title="Distribución del promedio semanal de productos por cliente",
    labels={"promedio_productos_semana": "Promedio de productos comprados por semana"}
)

fig.update_layout(
    bargap=0.1,
    xaxis_title="Promedio de productos/semana",
    yaxis_title="Número de clientes"
)

fig.show()
```

```python
transacciones_cliente = df.groupby('customer_id')['order_id'].nunique().reset_index(name='total transacciones')
transacciones_cliente
```

```python
fig = px.histogram(
    transacciones_cliente,
    x="total transacciones",
    nbins=30,  # número de bins (ajusta según prefieras)
    title="Distribución de total de transacciones por cliente",
    labels={"total transacciones": "Total de transacciones por cliente"}
)

fig.update_layout(
    bargap=0.1,
    xaxis_title="Numero de Productos",
    yaxis_title="Cantidad de clientes"
)

fig.show()
```

```python
# Ordenar por producto y fecha
df_sorted = df.sort_values(['product_id', 'purchase_date'])

# Calcular diferencia de días entre compras consecutivas por producto
df_sorted['diff_days'] = df_sorted.groupby('product_id')['purchase_date'].diff().dt.days

# Promedio de días de recompra por SKU (product_id)
recompra_promedio = df_sorted.groupby('product_id')['diff_days'].mean().reset_index(name='periodo_recompra_promedio')

recompra_promedio
```

```python

fig = px.histogram(
    recompra_promedio,
    x="periodo_recompra_promedio",
    nbins=40,  # ajusta el número de bins según distribución
    title="Distribución del periodo de recompra promedio por SKU",
    labels={"periodo_recompra_promedio": "Periodo de recompra promedio (días)"}
)

fig.update_layout(
    bargap=0.1,
    xaxis_title="Días entre compras",
    yaxis_title="Número de productos (SKU)"
)

fig.show()
```

En esta sección, se deben construir pipelines para automatizar el preprocesamiento de los datos, lo cual garantizará que el flujo de trabajo sea reproducible y eficiente para esta entrega y las futuras. El objetivo es aplicar una serie de transformaciones en un orden definido para asegurar que los datos estén listos para los modelos a entrenar. El pipeline final debe incluir las técnicas de pre-procesamiento que se deben aplicar a los distintos datos (según lo que consideren necesario para el problema). Por ejemplo:

- **Imputación de valores nulos**: Manejo de datos faltantes mediante estrategias adecuadas (media, mediana, moda, interpolación, etc.). 

- **Transformaciones personalizadas**: Uso de ColumnTransformer para aplicar diferentes transformaciones a columnas específicas.

- **Codificación de variables categóricas**: Convertir datos categóricos a un formato numérico adecuado (One-Hot Encoding, Label Encoding, etc.).

- **Discretización de variables**: Conversión de variables numéricas continuas en categorías si son relevantes para el desempeño del modelo a entrenar.

- **Estandarización o normalización** : Ajustar la escala de los datos para mejorar el rendimiento de los algoritmos sensibles a la magnitud de las variables.

- **Eliminación o transformación de valores atípicos**: Identificar y tratar con datos outliers para mejorar la robustez del modelo.

- **Nuevas características** : Creación de variables adicionales que puedan aportar información relevante al modelo.

Cada una de estas transformaciones debe ser justificada en función de su relevancia para el problema y los datos, y es importante evaluar su impacto en el rendimiento del modelo. Además, el pipeline debe ser flexible y modular para poder probar diferentes configuraciones de preprocesamiento.

```python
# Imputacion de valores nulos -> estrategias: media, mediana, moda, interpolación, etc
# Importamos las librerias necesarias:
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import FunctionTransformer
from sklearn.base import BaseEstimator, TransformerMixin
from sklearn.preprocessing import StandardScaler, OneHotEncoder, KBinsDiscretizer
from sklearn.impute import KNNImputer
import numpy as np

# Se establecen los tipos de las variables para luego asignar en los Pipelines correspondientes
num_features = ["payment", "X", "Y", "num_deliver_per_week", "num_visit_per_week", "size"]
num_features_dic = [] # Dependiendo del modelo a utilizar, esta lista se llenaría con las posibles variables a discretizar
cat_features = ["customer_type", "brand", "category", "sub_category", "segment", "package", "region_id"]

#Para crear la variable objetivo
class Obj(BaseEstimator,TransformerMixin):
  """Crea variable objetivo a partir de una la informacion del datset."""
  def __init__(self, df):
      self.df = df
    
  def fit(self, X, y=None):
      X = X.copy()
      return self
  
  def transform(self, X):
      X = X.copy()
      return X

# Para tratar con las fechas, haremos una transformación personalizada, esto ya que no consideramos correcto agregarlas a number 
# o category debido a su naturaleza -> no las estandarizamos y se usa one-hot encoding.
class Date_FeaturesExtractor(BaseEstimator, TransformerMixin):
    """Extrae variables de fecha a partir de una columna datetime."""
    def __init__(self, date_col):
        self.date_col = date_col
    
    def fit(self, X, y=None):# se calcula moda en fit para evitar dataleakage
        X = X.copy()
        self.nones = pd.to_datetime(X[self.date_col], errors="coerce").mode()[0]
        return self
    
    def transform(self, X):
        X = X.copy()
        X[self.date_col] = pd.to_datetime(X[self.date_col], errors="coerce")
        X[self.date_col] = X[self.date_col].fillna(self.nones)
        X["year"] = X[self.date_col].dt.year
        X["month"] = X[self.date_col].dt.month
        X["day"] = X[self.date_col].dt.day
        X["weekday"] = X[self.date_col].dt.weekday
        X = X.drop(columns=[self.date_col])
        return X
# Para tratar outliers usaremos el rango intercuartílico ya que se considera robusto para esta tarea
class IQR(BaseEstimator, TransformerMixin):

  def __init__(self, l):
    self.l = l
    self.inferior = -np.inf
    self.superior = np.inf

  def fit(self, X):
    q1 = X.quantile(0.25)
    q3 = X.quantile(0.75)
    RIC = q3-q1
    self.inferior = q1-self.l*RIC
    self.superior = q3+self.l*RIC
    return self

  def transform(self, X):
    for col in X.columns:
        X[col] = X[(X[col] >= self.inferior[col]) & (X[col] <= self.superior[col])][col]
    return X
      

  def set_output(self,transform='default'):
    return self
  
categoric_transformations = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")), # imputamos con las categorías más frecuentes -> porque son más probables
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

# Pipeline con discretización, sin escalado
numeric_transformations_disc = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=3)), # Usamos KNN porque en el lab fue el que mejores resultados dio
    ("scaler", StandardScaler()), # debido a que un gran cantidad de algoritmos de aprendizaje automático
     # / estadístico, asumen que los datos a operar se encuentran distribuidos de manera normal
    ("outlier", IQR(1.5)),  # Para outliers, 1.5 es un lambda bastante adecuado, 
    # ya que es el utilizado en los boxplots al momento de calcular los outliers.
    ("discretizer", KBinsDiscretizer(n_bins=5, encode="ordinal", strategy="quantile")) # en el caso de querer discretizar
])
# Pipeline sin discretización, con escalado
numeric_transformations_esc = Pipeline(steps=[
    ("imputer", KNNImputer(n_neighbors=3)), # Usamos KNN porque en el lab fue el que mejores resultados dio
    ("scaler", StandardScaler()), # debido a que un gran cantidad de algoritmos de aprendizaje automático
     # / estadístico, asumen que los datos a operar se encuentran distribuidos de manera normal
    ("outlier", IQR(1.5)),  # Para outliers, 1.5 es un lambda bastante adecuado, 
    # ya que es el utilizado en los boxplots al momento de calcular los outliers.
    ("scaler", StandardScaler()) # normalizamos para que más modelos puedan desempeñarse bien
])
```

```python
# Transformaciones personalizadas->sin discretización:
column_transformer_sin_discretizacion = ColumnTransformer(transformers=[
    ("num", numeric_transformations_esc, num_features),
    ("cat", categoric_transformations, cat_features)
])

pipeline = Pipeline(steps=[
    ("date_features", Date_FeaturesExtractor(date_col="purchase_date")),# Creamos variables
    ("preprocessor", column_transformer_sin_discretizacion)
])
```
